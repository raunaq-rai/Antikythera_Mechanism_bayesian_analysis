## **Coursework Overview**
This coursework involves Bayesian inference and Hamiltonian Monte Carlo (HMC) to analyze the Antikythera mechanism’s calendar ring. We aim to infer the original number of holes in the ring and assess competing statistical models.

The coursework is structured as follows:

- **(a)** Data acquisition and visualization
- **(b)** Model specification
- **(c)** Likelihood function definition
- **(d)** Log-likelihood gradient calculation
- **(e)** Maximum likelihood estimation (MLE)
- **(f)** Bayesian inference via HMC
- **(g)** Model comparison and interpretation

---

## **(a) Data Acquisition and Visualization** (4 marks)
- Download the data from the provided repository: [Replication Data](https://doi.org/10.7910/DVN/VJGLVS).
  - Two X-ray images of the Antikythera mechanism’s calendar ring.
  - A data file containing measured hole locations.
  - A README file (read carefully for dataset details).
- Plot the measured hole locations \( \mathbf{d}_i \in \mathbb{R}^2 \) in the \( x \)-\( y \) plane.
- Clearly indicate which holes belong to each fractured section of the ring.

## **(b) Model Specification** (5 marks)
- Use the model for hole locations from [Reference 2](https://arxiv.org/abs/2403.00040).
- Assume the complete ring originally contained \( N \) holes, arranged evenly around a circle of radius \( r \).
- The ring has fractured into 8 sections (\( j \in \{0,1,2,...,7\} \)), which may be misaligned due to:
  - **Translations** \( (x_j, y_j) \) in the \( x \)-\( y \) plane.
  - **Rotations** \( \alpha_j \).
- Treat \( N \) as a continuous parameter, allowing for the possibility of discontinuous hole spacing.

---

## **(c) Likelihood Function Definition** (5 marks)
- Use a **Gaussian likelihood function** for each hole location:

$$
L(D | \boldsymbol{\theta}) = \prod_{i=1}^{N_{\text{holes}}} \frac{1}{2\pi |\Sigma|} \exp \left( -\frac{1}{2} (\mathbf{d}_i - \mathbf{m}_i)^T \Sigma^{-1} (\mathbf{d}_i - \mathbf{m}_i) \right)
$$

where:
  - \( \mathbf{d}_i \) = measured hole locations.
  - \( \mathbf{m}_i \) = model-predicted hole locations.
  - \( \Sigma \) = covariance matrix representing uncertainty.
  - \( N_{\text{holes}} = 81 \) (number of measured holes in the surviving section).

- Consider **two models for \( \Sigma \)**:
  1. **Isotropic Gaussian model**: A 2D Gaussian with a single standard deviation \( \sigma \).
  2. **Radial/Tangential Gaussian model**: A 2D Gaussian with different standard deviations along radial (\( \sigma_r \)) and tangential (\( \sigma_t \)) directions.

- Define model parameters:
  - \( r \) (ring radius)
  - \( N \) (original number of holes)
  - \( \sigma \) or \( (\sigma_r, \sigma_t) \) (uncertainty parameters)
  - \( (x_j, y_j, \alpha_j) \) for each fragment \( j \).

## **(d) Log-Likelihood Gradient Calculation** (10 marks)
- Implement the gradient of the log-likelihood function:

$$
\frac{\partial}{\partial \theta_\mu} \log L(D | \boldsymbol{\theta})
$$

for each parameter \( \theta_\mu \).
- Provide details on how derivatives were calculated:
  - **Analytically** (by manual differentiation).
  - **Using automatic differentiation** (e.g., `jax` or `autograd` in Python).
- Verify the implementation with appropriate tests.

## **(e) Maximum Likelihood Estimation (MLE)** (8 marks)

  - Isotropic covariance model.
  - Radial/tangential covariance model.

  - Add markers for **maximum likelihood predictions** from both models.

## **(f) Bayesian Inference via Hamiltonian Monte Carlo (HMC)** (8 marks)
- Use **Hamiltonian Monte Carlo (HMC)** to sample from the posterior distribution of parameters.
- Compare results for both models.
- Pick one specific hole and:
  - Show the **posterior predictive distribution** for its location.
  - Visualize these posterior samples on the plot.

## **(g) Model Comparison and Interpretation** (10 marks)
- Explain the role of the **covariance matrix \( \Sigma \)** in the likelihood.
- Discuss why the **radial/tangential covariance model** is a reasonable assumption:
  - The holes were punched into a circular ring, so errors may be larger in certain directions.
- Determine if **data supports the radial/tangential model over the isotropic model**:
  - Use **quantitative metrics** (e.g., log-likelihood, Bayesian Information Criterion).
  - Provide **visual evidence** (plots showing model fit).
